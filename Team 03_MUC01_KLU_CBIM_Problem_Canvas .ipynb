# CBIM Problem Canvas

## Situation (S)
### Describe the retail chain and the upcoming supplier review in 1–2 sentences. What exists today?
A regional retail chain operates 12 stores across Andhra Pradesh and sells products across five categories: Electronics, Apparel, Grocery, Home & Kitchen, and Personal Care. The Category Manager is preparing for a quarterly supplier review and needs data-backed insights to evaluate category and product performance.

## Complication (C)
### What problem or gap does the Category Manager face going into the meeting?
The Category Manager needs to identify which categories are growing or declining, which products are driving or weakening revenue, and whether discounts are generating enough additional sales to justify their impact on revenue and margins.

## Question (Q)
### What is the specific business question this analysis must answer? Not "analyse the data" — what decision does it enable?
Based on six months of transaction data, which categories and products are performing strongly or weakly, how effective are discounts, and what should the Category Manager prioritise or challenge during the supplier review?

## SSOT
### What is the single source of truth for this analysis?
The single source of truth is the MUC01_Retail_Sales_Dataset.csv transaction-level sales dataset covering January to June 2026.

## Success Definition
### How will you know this analysis was useful? What does a good output look like in the supplier meeting?
The analysis will be useful if it provides clear category trends, identifies the strongest and weakest products, quantifies discount effectiveness, identifies useful sales timing patterns, and supports a specific data-backed discussion with suppliers.

## Primary Stakeholder
### Who is this analysis for? What does she need to walk away with?
The primary stakeholder is the Category Manager. She needs to leave the analysis with clear evidence about which category deserves greater attention, which supplier or product needs to be challenged, when promotions may be most useful, and what additional information is needed before making a final decision.


**Analysis principle:** Every visualization should answer a business question, not simply display a number.
